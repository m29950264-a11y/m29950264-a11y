```python
# Готовый код для Марии: Расчёт капитала к 30 годам
age = 19
target = 1000000
current_savings = 50000  # Допустим, стартовый грант или накопления
yearly_deposit = 12000   # Если откладывать по 1000 евро в месяц (12 000 в год)
interest_rate = 0.08     # 8% годовых — средняя доходность мирового рынка

# Цикл, который рассчитывает рост денег за 11 лет (с 19 до 30 лет)
for year in range(1, 12):
    current_savings = (current_savings + yearly_deposit) * (1 + interest_rate)
    current_age = age + year
    print(f"В {current_age} лет твой капитал: {round(current_savings):,} евро")

print("\n--- ИТОГ ---")
if current_savings >= target:
    print("Поздравляю, Мария! Цель в 1,000,000 евро ДОСТИГНУТА! 🚀")
else:
    print(f"Отличный результат! К 30 годам у тебя будет {round(current_savings):,} евро.")
    print(f"До еврового миллиона не хватает всего {round(target - current_savings):,} евро. Нужно наращивать доходы в AI!")
```

```python
# Исправленный код для Марии: Урок 2 (Библиотека Pandas)
import pandas as pd

# Полная база данных клиентов канадского стартапа
data = {
    'Client_Name': ['John', 'Anna', 'Maria', 'David', 'Elena'],
    'Country': ['Canada', 'Germany', 'Canada', 'France', 'Canada'],
    'Total_Spend_Euro': [12000, 45000, 95000, 23000, 51000],
    'Subscription_Years': [2, 4, 5, 1, 3]
}

# Переводим данные в таблицу Pandas
df = pd.DataFrame(data)

print("--- ПОЛНАЯ БАЗА ДАННЫХ КОМПАНИИ ---")
print(df)

# Автоматический подсчет среднего чека
average_spend = df['Total_Spend_Euro'].mean()
print(f"\nСредний чек клиента: {average_spend} евро")

# Находим VIP-клиентов, потративших больше 40 000 евро
vip_clients = df[df['Total_Spend_Euro'] > 40000]
print("\n--- VIP КЛИЕНТЫ С ВЫСОКИМ ДОХОДОМ ---")
print(vip_clients)
```

```python
import pandas as pd

# База данных клиентов канадского стартапа
data = {
    'Client_Name': ['John', 'Anna', 'Maria', 'David', 'Elena'],
    'Country': ['Canada', 'Germany', 'Canada', 'France', 'Canada'],
    'Total_Spend_Euro': [32000, 15000, 85000, 12000, 48000],
    'Subscription_Years': [2, 1, 5, 1, 3]
}

# Переводим данные в таблицу Pandas
df = pd.DataFrame(data)

print("--- ПОЛНАЯ БАЗА ДАННЫХ КОМПАНИИ ---")
print(df)

# Вычисляем средний чек
average_spend = df['Total_Spend_Euro'].mean()
print(f"\nСредний чек клиента: {average_spend} евро")

# Находим VIP-клиентов, потративших больше 40 000 евро
vip_clients = df[df['Total_Spend_Euro'] > 40000]
print("\n--- VIP КЛИЕНТЫ С ВЫСОКИМ ДОХОДОМ ---")
print(vip_clients)
```
